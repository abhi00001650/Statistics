Q1) What is a random variable in probability theory?

It is a function that assigns a numerical value to each outcome of a random experiment. It lets us represent uncertain events with numbers so we can analyze probabilities and compute quantities like expectations and variances.

Q2) What are the types of random variables?

Two types:

Discrete random variables - It takes conutable values ,  like the number of heads in coin tosses.

Continuous random variables - It takes values from a continuum or interval , like height , time or temperature.

Q3) Explain the difference between discrete and continuous distributions?

A discrete random variable takes countable, separate values like 0, 1 or 2 heads in coin tosses.

A continuous random variable takes values from an entire interval, meaning infinitely many possible values—like time, weight, or height. Discrete variables use probability mass functions , while continuous variables use probability density functions.

Q4)  What is a binomial distribution, and how is it used in probability?

A binomial distribution gives the probability of getting a certain number of successes in a fixed number of independent trials with a constant success probability. It is used for situations like counting heads in coin tosses or defects in a batch.

Q5) What is the standard normal distribution, and why is it important?

The standard normal distribution is a normal distribution with a mean of 0 and a standard deviation of 1. It is important because it allows us to standardize any normal variable, making probability calculations and comparisons easier using standard Z-scores and tables.

Q6) What is the Central Limit Theorem (CLT), and why is it critical in statistics?

The Central Limit Theorem says that if you take a large enough sample, the average of the sample will be roughly normal, no matter what the original data looks like. This helps us make conclusions about the population using the normal distribution.

Q7) What is the significance of confidence intervals in statistical analysis?

Confidence intervals show a range of values within which a population parameter is likely to lie, with a certain level of confidence like 95%. They help quantify uncertainty and provide more information than a single estimate, making statistical conclusions more reliable.

Q8) What is the concept of expected value in a probability distribution?

The expected value is the long-term average or mean of a random variable if an experiment is repeated many times. It represents the center of the probability distribution and helps predict the average outcome over the long run.

```python
#Q9) Write a Python program to generate 1000 random numbers from a normal
#distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
#deviation using NumPy, and draw a histogram to visualize the distribution.
#(Include your Python code and output in the code box below.)


import numpy as np
import matplotlib.pyplot as plt

mean = 50
std_dev = 5
num_samples = 1000

data = np.random.normal(loc = mean, scale= std_dev, size=num_samples)

calculated_mean = np.mean(data)
calculated_std_dev = np.std(data)

print("Mean:", calculated_mean)
print("Standard Deviation:", calculated_std_dev)

plt.hist(data, bins=30, edgecolor='black' , color = 'red')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Random Numbers')
plt.show()


```

```python
#Q10) You are working as a data analyst for a retail company. The company has
#collected daily sales data for 2 years and wants you to identify the overall sales trend.
#daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
#235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
#● Explain how you would apply the Central Limit Theorem to estimate the average sales
#with a 95% confidence interval.
#● Write the Python code to compute the mean sales and its confidence interval.
#(Include your Python code and output in the code box below.)


import numpy as np
from scipy import stats

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

n = len(daily_sales)
mean_sales = np.mean(daily_sales)
std_dev_sales = np.std(daily_sales, ddof=1)

confidence_level = 0.95
z_score = stats.norm.ppf(1 - (1 - confidence_level) / 2)
margin_of_error = z_score * (std_dev_sales / np.sqrt(n))

ci_lower = mean_sales - margin_of_error
ci_upper = mean_sales + margin_of_error

print(f"Mean Sales: {mean_sales:.2f}")
print(f"Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")


```
