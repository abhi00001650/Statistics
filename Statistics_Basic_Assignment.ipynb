```python
#Question 5: Implement a Python program to compute the mean, median, and mode of
#a given list of numbers.
#numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
#(Include your Python code and output in the code box below.)


import statistics as stats

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
mean_value = stats.mean(numbers)
median_value = stats.median(numbers)
mode_value = stats.mode(numbers)

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
#Question 6: Compute the covariance and correlation coefficient between the following
#two datasets provided as lists in Python:
#list_x = [10, 20, 30, 40, 50]
#list_y = [15, 25, 35, 45, 60]
#(Include your Python code and output in the code box below.)


import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

cov_value = np.cov(list_x, list_y , bias= False)
covaricance = cov_value[0,1]

corr_value = np.corrcoef(list_x, list_y)[0][1]

print("Covariance:", covaricance)
print("Correlation Coefficient:", corr_value)
```

```python
#Question 7: Write a Python script to draw a boxplot for the following numeric list and
#identify its outliers. Explain the result:
#data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
#(Include your Python code and output in the code box below.)


import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

plt.boxplot(data)
plt.title("Boxplot of Data")
plt.ylabel("Values")
plt.show()

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:" , Q1)
print("Q3:" ,Q3)
print("IQR" , IQR)
print("Outliers" , outliers)
```

The above box plot shows that  most data points are within the noraamal range but 35 lies far above the upper whisker making it an outlier

```python

#Question 8: You are working as a data analyst in an e-commerce company. The
#marketing team wants to know if there is a relationship between advertising spend and
#daily sales.
#● Explain how you would use covariance and correlation to explore this
#relationship.
#● Write Python code to compute the correlation between the two lists:
#advertising_spend = [200, 250, 300, 400, 500]
#daily_sales = [2200, 2450, 2750, 3200, 4000]
#(Include your Python code and output in the code box below.)


# Answer:


# Covariance and correlation help determine the relationship between advertising spend and daily sales.
# Covariance shows the direction of the relationship (positive or negative).
# Correlation measures both the strength and direction (values between -1 and 1).
# A positive correlation means higher advertising spend is associated with higher sales.

# Python code to compute correlation

import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Calculate covariance
cov_matrix = np.cov(advertising_spend, daily_sales, bias=False)
covariance = cov_matrix[0][1]

# Calculate correlation coefficient
correlation = np.corrcoef(advertising_spend, daily_sales)[0][1]

# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

The positive covairance and a correlation coefficient close to 1 show a strong positive relatiomship as advertising spend increases daily sales increases

```python
#Question 9: Your team has collected customer satisfaction survey data on a scale of
#1-10 and wants to understand its distribution before launching a new product.
#● Explain which summary statistics and visualizations (e.g. mean, standard
#deviation, histogram) you’d use.
#● Write Python code to create a histogram using Matplotlib for the survey data:
#survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
#(Include your Python code and output in the code box below.)

#Explanation

#To understand the distribution of customer satisfaction scores, you should use:

#Summary Statistics:

#Mean: Average satisfaction level.

#Median: Middle value — useful if the data is skewed.

#Mode: Most common rating — indicates the typical customer experience.

#Standard Deviation: Measures variability — how spread out the ratings are.

#Minimum and Maximum: Show the range of satisfaction.

#Visualizations:

#Histogram: Displays how frequently each satisfaction score occurs (good for visualizing distribution).

#Boxplot (optional): Shows spread, median, and potential outliers.

#Python Code


import matplotlib.pyplot as plt
import numpy as np


survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]


mean_score = np.mean(survey_scores)
median_score = np.median(survey_scores)
mode_score = max(set(survey_scores), key=survey_scores.count)
std_dev = np.std(survey_scores)

print(f"Mean: {mean_score:.2f}")
print(f"Median: {median_score}")
print(f"Mode: {mode_score}")
print(f"Standard Deviation: {std_dev:.2f}")


plt.hist(survey_scores, bins=range(1, 12), edgecolor='black', alpha=0.7)
plt.title('Customer Satisfaction Score Distribution')
plt.xlabel('Satisfaction Score (1–10)')
plt.ylabel('Frequency')
plt.xticks(range(1, 11))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()



```

The histogram would show:
Most customers rated between 6 and 9,
Centered around 7,
Indicating generally high satisfaction with a moderate spread.
