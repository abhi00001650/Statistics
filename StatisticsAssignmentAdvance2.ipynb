Question 1: What is hypothesis testing in statistics?

Hypothesis testing is a way to check if an idea about a population is likely true using sample data. It helps decide whether the evidence from the data supports the idea or not.

Question 2: What is the null hypothesis, and how does it differ from the alternative
hypothesis?

Null hypothesis: The no change or no effect idea.

Alternative hypothesis: The idea that there is an effect or difference.

Difference: H0 assumes nothing is happening ,  H1 says something is happening

Question 3: Explain the significance level in hypothesis testing and its role in deciding
the outcome of a test.

The significance level is the chance of rejecting the null hypothesis when it is actually true. It helps decide if the test result is strong enough to reject H0.

If the p-value ≤ a → reject H0

If the p-value > a → fail to reject H0

Question 4: What are Type I and Type II errors? Give examples of each.

Type I error: Rejecting the null hypothesis when it is true.

Example: A doctor says a patient has a disease when they actually don’t.

Type II error: Failing to reject the null hypothesis when it is false .

Example: A doctor says a patient is healthy when they actually have a disease.

Question 5: What is the difference between a Z-test and a T-test? Explain when to use
each.

Z-test: Used when the population standard deviation is known and/or the sample size is large usually >30.

T-test: Used when the population standard deviation is unknown and the sample size is small usually ≤30.

Z-test → big sample

T-test → small sample

```python
#Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in
#Python. Show the Python code and interpret the results.
#sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
#                        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
#                        50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
#                        50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
#(Include your Python code and output in the code box below.)


import numpy as np
from scipy import stats

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
                        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
                        50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
                        50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

mu = 50

x_bar = np.mean(sample_data)
sigma = np.std(sample_data , ddof = 0)
n = len(sample_data)


z = (x_bar - mu) / (sigma/np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"Sample mean: {x_bar:.2f}")
print(f"Z-statistc: {z:.2f}")
print(f"P-value: {p_value:.2f}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#Question 8: Write a Python script to simulate data from a normal distribution and
#calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
#(Include your Python code and output in the code box below.)


import numpy as np
import matplotlib.pyplot as plt
from scipy import stats


np.random.seed(0)
data = np.random.normal(loc=50, scale=5, size=100)


x_bar = np.mean(data)
se = np.std(data, ddof=1) / np.sqrt(len(data))


confidence = 0.95
ci = stats.t.interval(confidence, df=len(data)-1, loc=x_bar, scale=se)

print(f"Sample mean: {x_bar:.2f}")
print(f"95% confidence interval: {ci}")


plt.hist(data, bins=15, edgecolor='black', alpha=0.7)
plt.axvline(x_bar, color='red', linestyle=':', linewidth=2, label='Mean')
plt.title("Normal Distribution Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
#Question 9: Write a Python function to calculate the Z-scores from a dataset and
#visualize the standardized data using a histogram. Explain what the Z-scores represent
#in terms of standard deviations from the mean.
#(Include your Python code and output in the code box below.)

import numpy as np
import matplotlib.pyplot as plt

def z_scores(data):
    mean = np.mean(data)
    std = np.std(data, ddof=0)
    z = (data - mean) / std
    return z


data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5]


z = z_scores(np.array(data))

print("Z-scores:", np.round(z, 2))


plt.hist(z, bins=10, edgecolor='black', alpha=0.7)
plt.title("Z-scores of the Dataset")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.show()
```
